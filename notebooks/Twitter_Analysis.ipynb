## Creating Time Features ##

```python
df["date"] = df["time"].dt.date
df["year"] = df["time"].dt.year
df["month"] = df["time"].dt.month
df["month_name"] = df["time"].dt.month_name()
df["day"] = df["time"].dt.day
df["day_of_week"] = df["time"].dt.day_name()
df["hour"] = df["time"].dt.hour
df.head(10)
```

## Calculate Interactions ##

```python
interaction_cols = [
    "likes",
    "retweets",
    "replies",
    "user profile clicks",
    "url clicks",
    "hashtag clicks"
]

df["calculated_interactions"] = df[interaction_cols].sum(axis=1)

df[[
    "engagements",
    "calculated_interactions"
]].head(20)
```

## Engagement Difference ##

```python
difference = (
    df["engagements"] - df["calculated_interactions"]
)
print(difference.describe())

df["interaction_difference"]=difference

df.nlargest(
    10,
    "interaction_difference"
)[[
    "Tweet",
    "time",
    "engagements",
    "calculated_interactions",
    "interaction_difference"
]]
```

### Impressions Distribution ###

```python
impression_bins = pd.cut(
    df["impressions"],
    bins=[0, 100, 500, 1000, 5000, 10000, float("inf")],
    labels=[
        "0–100",
        "101–500",
        "501–1K",
        "1K–5K",
        "5K–10K",
        "10K+"
    ]
)

impression_distribution = (
    impression_bins
    .value_counts()
    .sort_index()
)

impression_distribution
```

### Engagement Distribution ###

```python
engagement_bins = pd.cut(
    df["engagements"],
    bins=[-1, 5, 10, 25, 50, 100, 500, 1000, float("inf")],
    labels=[
        "0–5",
        "6–10",
        "11–25",
        "26–50",
        "51–100",
        "101–500",
        "501–1K",
        "1K+"
    ]
)

engagement_distribution = (
    engagement_bins
    .value_counts()
    .sort_index()
)

engagement_distribution
```

### Interaction Summary ###

```python
interaction_cols = [
    "likes",
    "retweets",
    "replies",
    "user profile clicks",
    "url clicks",
    "hashtag clicks"
]

interaction_summary = pd.DataFrame({
    "Total": df[interaction_cols].sum(),
    "Average": df[interaction_cols].mean(),
    "Median": df[interaction_cols].median()
})

interaction_summary["Percentage"] = (
    interaction_summary["Total"]
    / interaction_summary["Total"].sum()
    * 100
)

interaction_summary.sort_values(
    "Total",
    ascending=False
)
```

## Time Based Analysis ##

```python
monthly_performance = (
    df.groupby("month_name", sort=False)
      .agg(
          tweets=("Tweet", "count"),
          impressions=("impressions", "sum"),
          engagements=("engagements", "sum"),
          avg_engagement_rate=("engagement rate", "mean")
      )
)

monthly_performance
```

```python
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_performance = (
    df.groupby("day_of_week")
      .agg(
          tweets=("Tweet", "count"),
          impressions=("impressions", "sum"),
          engagements=("engagements", "sum"),
          avg_engagement_rate=("engagement rate", "mean")
      )
      .reindex(day_order)
)
day_performance["avg_impressions_per_tweet"] = (
day_performance["impressions"] / day_performance["tweets"]
)

day_performance["avg_engagements_per_tweet"] = (
day_performance["engagements"] / day_performance["tweets"]
)

day_performance
```

```python
friday_tweets = df[df["day_of_week"] == "Friday"].copy()
friday_tweets[
    ["Tweet", "impressions", "engagements", "engagement rate"]
].sort_values(
    "engagements",
    ascending=False
).head(10)
```

```python
hour_analysis = df.groupby("hour").agg(
    tweets=("id", "count"),
    impressions=("impressions", "sum"),
    engagements=("engagements", "sum"),
    avg_engagement_rate=("engagement rate", "mean"),
    median_engagement_rate=("engagement rate", "median")
).reset_index()

hour_analysis
```

```python
day_hour_analysis = df.groupby(
    ["day_of_week", "hour"]
).agg(
    tweets=("id", "count"),
    impressions=("impressions", "sum"),
    engagements=("engagements", "sum"),
    avg_engagement_rate=("engagement rate", "mean"),
    median_engagement_rate=("engagement rate", "median")
).reset_index()

day_hour_analysis
```
